```python
# statistics

a = [[1,2,3],[4,5,6],[7,8,9]]
a = np.array(a)
a
```

```python
print(np.min(a,axis=0))
print(np.min(a,axis=1))
print(np.sum(a))
```

```python
# rearange

print(a.reshape(1,9))
print(a.reshape(9,1))
```
