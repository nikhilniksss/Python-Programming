```python
# convert anything with format parameter

dates = ['2025$01$01','2025$08$27']
pd.to_datetime(dates,format="%Y$%m$%d")
```

```python
# convert anything with format parameter

dates = ['2025/01/01','2025/08/27']
pd.to_datetime(dates,format="%Y/%m/%d")
```

```python
# Handling two or more formats

dates = ['2025/01/01','2025-08-27']
pd.to_datetime(dates,format="mixed")
```

```python
# while lodaing the data we can use the set index to make the index value

df = pd.read_csv("/Users/nick_mac/Desktop/python/namastepython/orders.txt",parse_dates=['order_date'],index_col='order_date')
df.head()
```

```python
# set index using year of the data

df.loc['2020']
```

```python
# data slicing

df.sort_index().loc['2020-01-01':'2020-01-05']
```

```python
# get individual attributes of dates

df.reset_index(inplace=True)

print(df['order_date'].dt.day_name())
print(df['order_date'].dt.date)
print(df['order_date'].dt.weekday)
```

```python
# visualization using matplot lib

df.plot('category','sales')
```

```python
# change the type of charts

df.plot('category','sales',kind='bar')
```

```python
# time series visualizstion

df = pd.read_csv("/Users/nick_mac/Desktop/python/namastepython/orders.txt",parse_dates=['order_date'],index_col='order_date')
```

```python
df['sales'].plot()
```

```python
# category wise graph

df.groupby('category')['sales'].sum().plot()
```

```python
# category wise graph

df.groupby('category')['sales'].sum().plot(kind = 'barh')
```

```python
# year wise or date wise

df['sales'].resample('YE').sum()
```

```python
# year wise or date wise

df['sales'].resample('YE').sum().plot()
```

```python
# Quarter wise

df['sales'].resample('QE').sum().plot(title="Quarter wise sales")
```
