```python
# Q3- from a list of numbers create 2 list , one containing only the even numbers and other only the odd numbers

list1 = [2,4,1,3,7,6,9,3]
list_even = []
list_odd = []

for i in list1:
    if i % 2 == 0:
        list_even.append(i)
    else:
        list_odd.append(i)

print(list1)
print(list_even)
print(list_odd)
```
