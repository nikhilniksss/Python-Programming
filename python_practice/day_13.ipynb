```python
import pandas as pd
```

```python
emp = {'name':['Nikhil','Amit','Rahul'],
        'gender':['Male','Male','Male'],
        'email':['nikhil@gmail.com','amit@yahoo.com','rahul@reddif.mail']}
```

```python
# creating a dataframe

df_emp = pd.DataFrame(emp)
```

```python
df_emp
```

```python
# apply function

df_emp['name'].apply(len)
```

```python
# apply with user defined function

def update_email(email):
    return email.upper()
```

```python
df_emp['email'].apply(update_email)
```

```python
# applymap is for applying a similar function to entire dataframe

df_emp.applymap(len)
```

```python
df_emp.applymap(update_email)
```

```python
df_emp['name'].str.upper()
```

```python
# applying string functions

df_emp['name'].str.lower()
```

```python
# contains function

df_emp['name'].str.startswith('A')
```

```python
# Add new column

df_emp['Salary'] = 10000
```

```python
df_emp
```

```python
# new column with concatenate

df_emp['name_gender'] = df_emp['name']+' '+df_emp['gender']
```

```python
df_emp
```

```python
# delete rows

df_emp.drop(index=0) # dropping the rows which is at index 0
```

```python
# filter

df_emp['Salary'] > 1000
```

```python
# sort the dataframe

df_emp.sort_values(by='name') # by default it's in ascending
```

```python
# sort the dataframe

df_emp.sort_values(by='name',ascending=False) # by descending
```

```python
# to reset the index

df_emp.sort_index()
```

```python
df_emp.sort_index(ascending=False)
```
