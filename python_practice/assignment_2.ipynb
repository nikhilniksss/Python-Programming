3- given a list list1=[1,2,3,4,5,6,7,8]

diplay a new list which contains only odd position index values from above list.

```python
list1 = [1,2,3,4,5,6,7,8]
list2 = []
for i in list1:
    if list1.index(i)%2 != 0:
        list2.append(i)
print(list2)
```

8-ipl= ['CSK','MI','KKR','LSG','PBKS']

take a user input contains 2 comma seprated values index,new_team. 
Add the new value at that index in the list. 
Display the old list , new list,length of original and new list

example : input : 2,SRH
output : 
old list : ['CSK','MI','KKR','LSG','PBKS'] and length 5
new list : ['CSK','MI','SRH','KKR',LSG','PBKS'] and length 6

```python
ipl_old = ['CSK','MI','KKR','LSG','PBKS']
print(f"old list:{ipl_old} and length {len(ipl_old)}")

user_input = input("Enter a team:")
user_input = user_input.split(",")
ipl_new = ipl_old
ipl_new.insert(int(user_input[0]),user_input[1])

print(f"new list:{ipl_new} and length {len(ipl_new)}")
```
